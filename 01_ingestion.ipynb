```python
dbutils.widgets.text("catalog", "dev")
catalog = dbutils.widgets.get("catalog")

dbutils.widgets.text("job_run_id", "ADHOC")
job_run_id = dbutils.widgets.get("job_run_id")
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog}.bronze.passenger;
""")

spark.sql(f"""
COPY INTO {catalog}.bronze.passenger
FROM (SELECT *,
    _metadata.file_name as _source_file,
    current_timestamp() as _ingested_at,
    '{job_run_id}' as job_run_id
    from '/Volumes/dev/bronze/raw/passengers/'
    )
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
""")
```

```python
%sql
select count(*) from dev.bronze.passenger
```

```python
%sql
select count(*) from read_files("/Volumes/dev/bronze/raw/passengers/passenger.csv")
```
