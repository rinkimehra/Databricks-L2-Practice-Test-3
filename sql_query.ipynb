```python
%sql
-- Q01: KPI Summary
CREATE OR REPLACE VIEW dev.gold.vw_q01_kpi_summary AS
SELECT
  SUM(total_amount) AS total_net_revenue,
  COUNT(DISTINCT booking_id) AS distinct_bookings,
  COUNT(DISTINCT passenger_id) AS active_passengers,
  AVG(total_amount) AS average_booking_value
FROM dev.gold.fact_bookings;

-- Interpretation: High net revenue and active passenger count indicate strong booking trends. Management should focus on maintaining booking growth and monitoring value per booking.
-- Action: Review pricing, expand customer acquisition, adjust marketing channels.
```

```python
%sql
-- Q02: Monthly Revenue Trend
CREATE OR REPLACE VIEW dev.gold.vw_q02_monthly_revenue_trend AS
SELECT
  YEAR(booking_date) AS booking_year,
  MONTH(booking_date) AS booking_month,
  SUM(total_amount) AS monthly_net_revenue
FROM dev.gold.fact_bookings
GROUP BY booking_year, booking_month
ORDER BY booking_year, booking_month;

-- Interpretation: Reveals seasonality and growth patterns. Management should optimize inventory and pricing for peak and off-peak months.
-- Action: Align marketing and operations to monthly demand patterns.
```

```python
%sql
-- Q03: Annual Revenue With YoY Change
CREATE OR REPLACE VIEW dev.gold.vw_q03_annual_revenue_yoy AS
SELECT
  booking_year,
  net_revenue AS annual_net_revenue,
  LAG(net_revenue) OVER (ORDER BY booking_year) AS prior_year_revenue,
  ROUND((net_revenue - LAG(net_revenue) OVER (ORDER BY booking_year)) / NULLIF(LAG(net_revenue) OVER (ORDER BY booking_year), 0) * 100, 2) AS yoy_change_pct
FROM dev.gold.vw_annual_booking_kpis
ORDER BY booking_year;

-- Interpretation: Measures yearly revenue growth or contraction. Enables strategic planning and progress tracking.
-- Action: Investigate causes for decline or acceleration, target improvement initiatives.
```

```python
%sql
-- Q04: Revenue by Passenger Travel Purpose
CREATE OR REPLACE VIEW dev.gold.vw_q04_revenue_by_travel_purpose AS
SELECT
  trip_type AS travel_purpose,
  SUM(total_amount) AS revenue,
  COUNT(DISTINCT booking_id) AS booking_count
FROM dev.gold.fact_bookings
GROUP BY trip_type
ORDER BY revenue DESC;

-- Interpretation: Identifies business, leisure, or group travel segments. Management can tailor offerings to high-value segments.
-- Action: Adjust products, services, marketing by travel purpose segments.
```

```python
%sql
-- Q05: Top 10 Passengers by Net Revenue
CREATE OR REPLACE VIEW dev.gold.vw_q05_top10_passengers AS
SELECT
  p.passenger_name,
  f.passenger_id,
  SUM(f.net_revenue) AS passenger_revenue
FROM dev.gold.fact_bookings f
JOIN dev.gold.dim_passenger p ON f.passenger_id = p.passenger_id
GROUP BY p.passenger_name, f.passenger_id
ORDER BY passenger_revenue DESC
LIMIT 10;

-- Interpretation: Highlights key revenue-contributing customers.
-- Action: Engage top passengers, offer loyalty programs, targeted retention.
```
